# 3. تقسيم البيانات إلى Train, Validation و Test (Data Splitting)

```python
import pandas as pd

df = pd.read_csv('olist_with_labels.csv')
len(df)
df.head()
```

```python
n = len(df)
train_end = int(n * 0.70)      # أول 70% للتدريب
val_end = int(n * 0.85)        # الـ 15% التالية للتحقق (من 70% لـ 85%)
# الباقي (آخر 15%) رح يكون للـ Test
```

```python
train_df = df.iloc[:train_end]
val_df = df.iloc[train_end:val_end]
test_df = df.iloc[val_end:]
```

```python
print(f"Train rows: {len(train_df)}")
print(f"Validation rows: {len(val_df)}")
print(f"Test rows: {len(test_df)}")
```

```python
# حفظ الأقسام الثلاثة بملفات CSV مستقلة
train_df.to_csv('train.csv', index=False)
val_df.to_csv('val.csv', index=False)
test_df.to_csv('test.csv', index=False)

print("Successfully saved train.csv, val.csv, and test.csv!")
```
