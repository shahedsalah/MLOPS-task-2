# 2. إنشاء الليبل وتحديد مشكلة عدم توازن البيانات (Label Creation & Class Imbalance)

```python
import pandas as pd

df = pd.read_csv('olist_master_artifact.csv')
df.head()
```

```python
df['order_delivered_customer_date'] = df['order_delivered_customer_date'].astype('datetime64[ns]')
df['order_delivered_customer_date']
```

```python
df['order_estimated_delivery_date'] = df['order_estimated_delivery_date'].astype('datetime64[ns]')
df['order_estimated_delivery_date']
```

```python
df.info()
```

```python
df['is_late'] = (df['order_delivered_customer_date'] > df['order_estimated_delivery_date']).astype(int)
```

```python
# 0 the number of on-time deliveries
# 1 the number of late orders
df['is_late'].value_counts()
```

```python
# حفظ الداتا مع الليبل الجديد بملف CSV
df.to_csv('olist_with_labels.csv', index=False)

print("Successfully saved olist_with_labels.csv!")
```

# Class Imbalance Observation:
### The label distribution shows a significant class imbalance; on-time deliveries (0) account for approximately 92% of the dataset (91,614 orders), while delayed deliveries (1) account for only 8% (7,827 orders). This imbalance means that relying solely on Accuracy for model evaluation will be misleading, and we should consider using appropriate metrics (like Precision, Recall, and F1-Score) or handling techniques (such as class_weight) during the machine learning phase.
