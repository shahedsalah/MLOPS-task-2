```python
date_columns = [
    'order_purchase_timestamp', 
    'order_approved_at', 
    'order_delivered_carrier_date', 
    'order_delivered_customer_date', 
    'order_estimated_delivery_date'
]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col])



columns_to_drop = [
    'order_id', 
    'customer_id', 
    'customer_unique_id', 
    'review_id', 
    'review_comment_title', 
    'review_comment_message',
    'geolocation_zip_code_prefix',
    'customer_zip_code_prefix'
]

df = df.drop(columns=[col for col in columns_to_drop if col in df.columns], errors='ignore')

print(df.shape)
df.head()
    
```

## (is_business_hour) بقيم (0 أو 1) بناءً على إذا كانت ساعة الطلب تقع بين 8 صباحاً و 6 مساءً مثلاً.

```python
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])

df['date'] = df['order_purchase_timestamp'].dt.date
df['time'] = df['order_purchase_timestamp'].dt.time

df['order_hour'] = df['order_purchase_timestamp'].dt.hour

def get_time_of_day(hour):
    if 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 22:
        return 'Evening'
    else:
        return 'Night'

df['time_of_day'] = df['order_hour'].apply(get_time_of_day)


df['is_business_hour'] = df['order_hour'].between(8, 18).astype(int)

```

### approval_time_days: تبين كم تأخر المتجر بالموافقة على الطلب.

### processing_time_days: تبين كفاءة التجهيز وتسليمها للشحن.

### shipping_time_days: تبين سرعة شركة الشحن نفسها.

### delivery_vs_estimated_days: توضح ما إذا كان الطلب وصل مبكراً أم متأخراً عن الموعد المتوقع.

```python

# أ. وقت معالجة الطلب والموافقة عليه (من الشراء للموافقة)
df['approval_time_days'] = (df['order_approved_at'] - df['order_purchase_timestamp']).dt.total_seconds() / (24 * 3600)

# ب. وقت التجهيز والشحن للناقل (من الموافقة لتسليمها لشركة الشحن)
df['processing_time_days'] = (df['order_delivered_carrier_date'] - df['order_approved_at']).dt.total_seconds() / (24 * 3600)

# ج. وقت التوصيل الفعلي للزبون (من التسليم للناقل وحتى وصولها للزبون)
df['shipping_time_days'] = (df['order_delivered_customer_date'] - df['order_delivered_carrier_date']).dt.total_seconds() / (24 * 3600)

# د. إجمالي مدة التوصيل (من الشراء وحتى الاستلام الفعلي)
df['total_delivery_days'] = (df['order_delivered_customer_date'] - df['order_purchase_timestamp']).dt.total_seconds() / (24 * 3600)

# هـ. الفرق بين المتوقع والفعلي (مهم جداً لتوقع التأخير)
df['delivery_vs_estimated_days'] = (df['order_estimated_delivery_date'] - df['order_delivered_customer_date']).dt.total_seconds() / (24 * 3600)

```

```python
# الأعمدة التالية تحمل "معلومات من المستقبل" (Data Leakage) 
# ولا يجب إعطاؤها للموديل أثناء التدريب لكي يتوقع التأخير بحيادية ومنطقية:
columns_to_drop_for_leakage = [
    'delivery_vs_estimated_days',  # لأنها تحسب الفارق المباشر مع التاريخ الفعلي (تكشف النتيجة المطلوبة للموديل مسبقاً)
    'review_score',                # لأن تقييم الزبون يكتبه ويحدث بعد أن يستلم الطلب وينتهي كل شيء
    'shipping_time_days',          # لأن أيام الشحن الفعلية تعتمد على تاريخ الوصول الحقيقي للزبون (حدث انتهى بالماضي)
    'total_delivery_days'          # لأن إجمالي أيام التوصيل يحسب حتى تاريخ الاستلام الفعلي (معلومة لا تتوفر وقت بدء الطلب)
]

# حذف الأعمدة من الداتا فريم بأمان
df = df.drop(columns=[col for col in columns_to_drop_for_leakage if col in df.columns], errors='ignore')

```

```python
import numpy as np
import pandas as pd

# 1. تحديد الأعمدة العددية (باستثناء هدفنا is_late وأعمدة الـ One-Hot Encoding الثنائية 0 و 1)
numerical_cols = [
    'total_price', 'total_freight', 'total_items', 'total_weight_g', 
    'total_payment', 'max_installments', 'order_hour', 
    'approval_time_days', 'processing_time_days'
]

# التأكد من أن الأعمدة موجودة فعلاً في الداتا فريم
numerical_cols = [col for col in numerical_cols if col in df.columns]

# 2. فحص الالتواء (Skewness) لكل عمود عددي
skewness_check = df[numerical_cols].skew()
print("--- نسبة الالتواء لكل عمود ---")
print(skewness_check)

# 3. علاج الالتواء العالي باستخدام Log Transformation للأعمدة ذات الالتواء الشديد (مثلاً > 1)
for col in numerical_cols:
    if df[col].skew() > 1:
        # استخدام log1p (أي log(1 + x)) لتجنب مشاكل الصفر
        df[col] = np.log1p(df[col])
        print(f"تم تطبيق Log Transformation على العمود: {col}")

```

```python
import numpy as np

# 1. استبدال قيم ما لا نهاية (inf) بـ NaN
df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
from sklearn.preprocessing import StandardScaler

# 1. تحديد الأعمدة العددية التي تحتاج إلى Scaling (نستثني منها الأعمدة الثنائية 0 و 1 والهدف is_late)
scaling_cols = [
    'total_price', 'total_freight', 'total_items', 'total_weight_g', 
    'total_payment', 'max_installments', 'order_hour', 
    'approval_time_days', 'processing_time_days'
]

scaling_cols = [col for col in scaling_cols if col in df.columns]
df.dropna(subset=scaling_cols, inplace=True)

# 2. تطبيق StandardScaler
scaler = StandardScaler()
df[scaling_cols] = scaler.fit_transform(df[scaling_cols])
```
